```python
from contract_mapping_v2 import contract_mapper_v2
import pandas as pd
from pathlib import Path
from python_system_prompt import system_prompt, system_prompt_v2
```

```python
df = pd.read_csv(Path.cwd() / "AI Category Mapping - Category Desc Examples_new.csv") #should be a csv with a
""""
data should look like this :
|Category    |	Description |
|Construction|	City wall repairs|
the columns are Category and Description
"""
df = df.dropna(subset=['Description']).copy() #in case anyone forgets to add a description
df = df.reset_index(drop=True)
list_descriptions = df["Description"].values.tolist()
list_category = df["Category"].values.tolist()
```

```python
count_correct_v1 = 0
count_correct_v2 = 0
list_results_v1 = []
list_results_v2 = []

for description, category in zip(list_descriptions, list_category):
    result_v1 = contract_mapper_v2(user_contract_description=description, system_prompt=system_prompt())
    list_results_v1.append(result_v1)

    result_v2 = contract_mapper_v2(user_contract_description=description, system_prompt=system_prompt_v2())
    list_results_v2.append(result_v2)

    print(f"Old system prompt, actual:{category}  AI:{result_v1}, {category in result_v1}")
    print(f"Optimised system prompt, actual:{category}  AI:{result_v2}, {category in result_v2}")
    if category == result_v1:
        count_correct_v1 += 1

    if category == result_v2:
        count_correct_v2 += 1
    print()


print(f"Old system prompt AI accuracy:{(count_correct_v1/len(list_descriptions))*100}% on {len(list_results_v1)} samples")

print(f"Optimised system prompt AI accuracy:{(count_correct_v2/len(list_descriptions))*100}% on {len(list_results_v2)} samples")

```

```python
# save to csv
new_df = df[["Category", "Description"]].copy()
new_df["AI classification"] = list_results_v1

new_df.to_csv("AI_v2_results_old.csv")

new_df2 = df[["Category", "Description"]].copy()
new_df2["AI classification"] = list_results_v2

new_df2.to_csv("AI_v2_results_optimised.csv")

print("experiment completed")
```
