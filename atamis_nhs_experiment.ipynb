## Prepare data

make sure you get the data which you can obtain from the data team

```python
data = pd.read_excel('Davis_atamis_test.xlsx')
```

```python
# choosing the most relevant columns
nhs_contract_descr_data = data[["Primary Category Name", "Description"]]
```

## Run experiment using contract mapping on NHS

```python
labels = []
for index, row in nhs_contract_descr_data.iterrows():

    contract_description = f"Title: {row['Primary Category Name']}\n Description {row['Description']}"

    # We 'await' the result immediately
    label = await contract_mapper_v2(user_contract_description=contract_description)
    labels.append(label)

nhs_contract_descr_data["AI_label"] = labels
```

```python
#save new labels to original dataframe and save to csv
data["AI_label"] = labels
data.to_csv('all_data_NHS_atamis_contractmap_results.csv')
```
